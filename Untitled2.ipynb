## Soal 1: Prioritas Operator
Dalam bahasa pemrograman Python, operator perkalian (*) memiliki tingkat prioritas (precedence) yang lebih tinggi dibandingkan dengan operator penjumlahan (+).
Oleh karena itu, pada ekspresi matematika "10 + 5 * 2", Python akan mengeksekusi operasi perkalian terlebih dahulu yaitu 5 * 2 = 10, baru kemudian menjumlahkannya dengan 10 sehingga hasil akhirnya adalah 20. Jika ingin operasi penjumlahan didahulukan agar hasilnya menjadi 30, maka operasi tersebut harus dibungkus dengan tanda kurung menjadi "(10 + 5) * 2".

## Soal 2: Operator Relasional vs Logika
- Operator Relasional (seperti >, <, >=, <=, ==, !=) digunakan untuk membandingkan dua nilai atau operan. Hasil dari operasi relasional ini selalu berupa nilai Boolean (True atau False).
- Operator Logika (seperti and, or, not) digunakan untuk menghubungkan atau mengombinasikan dua atau lebih kondisi Boolean menjadi satu keputusan logis tunggal.
Contoh Kasus Nyata: Skenario pengecekan kelayakan beasiswa mahasiswa, di mana mahasiswa dinyatakan layak jika "IPK di atas 3.5" DAN "Skor TOEFL di atas 500". Bentuk ekspresi kodenya adalah: (ipk > 3.5) and (toefl > 500).

## Soal 3: Kasus Bug Prioritas Operator
Rangkuman Kasus: Sebuah bug terjadi pada sistem kalkulasi total belanja e-commerce saat kode ditulis sebagai "total = harga - diskon * jumlah_barang". Pengembang berniat memotong harga dengan diskon terlebih dahulu baru dikalikan jumlah barang. Namun, karena perkalian dihitung lebih dulu, nilai diskon dikalikan jumlah barang dahulu sebelum mengurangi harga, membuat total tagihan salah besar dan merugikan toko.
Solusi perbaikan kode yang benar dengan tanda kurung:
total = (harga - diskon) * jumlah_barang
