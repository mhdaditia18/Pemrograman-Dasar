```python
tahun = 2026
(tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
```

```python
# 1. Fungsi Input
nama_barang = input("Masukkan nama barang: ")
harga = int(input("Masukkan harga barang: "))
jumlah = int(input("Masukkan jumlah beli: "))
uang_dibayar = int(input("Masukkan uang tunai yang dibayar: "))

# 2. Operasi Aritmatika
total_belanja = harga * jumlah
kembalian = uang_dibayar - total_belanja

# 3. Fungsi Output & Logika Cukup/Tidak
print("\n--- NOTA PEMBAYARAN ---")
print("Barang         :", nama_barang)
print("Total Belanja  : Rp", total_belanja)
print("Uang Dibayar   : Rp", uang_dibayar)

if uang_dibayar >= total_belanja:
    print("Uang Kembalian : Rp", kembalian)
    print("Status         : Pembayaran Sukses")
else:
    kekurangan = total_belanja - uang_dibayar
    print("Status         : Uang Kurang Rp", kekurangan)
```
